Задание 1

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("vgsales.csv")
print("Первые 10 строк:")
df.head(10)
```

Задание 2

```python
print("\nРазмер таблицы:", df.shape)
print("\nСтолбцы:", df.columns.tolist())
print("\nТипы данных:")
print(df.dtypes)

print("\nСтатистика по числовым столбцам:")
print(df.describe())

print("\nЧисловые столбцы:")
print(df.select_dtypes(include=np.number).columns.tolist())

print("\nТекстовые столбцы:")
print(df.select_dtypes(include="object").columns.tolist())

sales_cols = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales", "Global_Sales"]
print("\nСтолбцы с продажами:", sales_cols)
```

Задание 3

```python
missing = df.isnull().sum()
print("\nПропуски по столбцам:")
print(missing)

print("\nСтолбцы с пропусками:")
print(missing[missing > 0])

print("\nБольше всего пропусков в столбце:")
print(missing.idxmax(), "—", missing.max())

print("\nСтроки, где отсутствует Year:")
print(df[df["Year"].isnull()])
```

Задание 4

```python
df_clean = df.dropna(subset=["Year"]).copy()

print("\nРазмер исходного df:", df.shape)
print("Размер df_clean:", df_clean.shape)

print("\nПропуски в df_clean:")
print(df_clean.isnull().sum())
```

*   всего строк в исходном датасете: 16598
*   После удаления пропусков Year: 16327
*   Пропуски в Year не позволяют отнести игру к конкретному году, поэтому её нельзя учесть при группировке, подсчёте статистики и построении трендов по годам — это искажает результаты анализа.

Задание 6 Анализ жанров

```python
genre_count = df_clean.groupby("Genre")["Name"].count().sort_values(ascending=False)
print("\nКоличество игр по жанрам:")
print(genre_count)
```

Задание 7 Продажи по жанрам

```python
genre_sales = df_clean.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False)
print("\nМировые продажи по жанрам:")
print(genre_sales)
```

Задание 9 Анализ платформ

```python
platform_count = df_clean.groupby("Platform")["Name"].count().sort_values(ascending=False)
print("\nКоличество игр для каждой платформы:")
print(platform_count.head(31))
print("\nТоп-10 платформ по количеству игр:")
print(platform_count.head(10))

platform_sales = df_clean.groupby("Platform")["Global_Sales"].sum().sort_values(ascending=False)
print("\nТоп-10 платформ по суммарным продажам:")
print(platform_sales.head(10))
```

Задание 10 Визуализация продаж по жанрам

```python
plt.figure(figsize=(12, 6))
genre_sales.sort_values(ascending=False).plot(kind="bar")
plt.title("Суммарные мировые продажи по жанрам")
plt.xlabel("Жанр")
plt.ylabel("Продажи, млн")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

Задание 11 Корреляция продаж

```python
corr_na_global = df_clean[["NA_Sales", "Global_Sales"]].corr()
print("\nКорреляция NA_Sales и Global_Sales:")
print(corr_na_global)
```

Корреляция NA_Sales и Global_Sales около 0.94.

Это положительное значение, которое близко к 1, значит, связь выраженная положительная. Но важно помнить: Global_Sales включает в себя NA_Sales, поэтому такая высокая корреляция ожидаема. Корреляция не доказывает причинно-следственную связь.

Задание 12 Диаграмма рассеяния

```python
plt.figure(figsize=(8, 5))
plt.scatter(df_clean["NA_Sales"], df_clean["Global_Sales"], alpha=0.5)
plt.title("Связь продаж в Северной Америке и мировых продаж")
plt.xlabel("Продажи в Северной Америке, млн")
plt.ylabel("Мировые продажи, млн")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

Одна точка — одна игра.
По графику видна восходящая тенденция: чем больше продажи в Северной Америке, тем обычно больше мировые продажи.
Это соответствует результату корреляционного анализа.

Задание 13 Корреляция между регионами

```python
region_corr = df_clean[["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]].corr()
print("\nКорреляционная таблица по регионам:")
print(region_corr)

# Поиск самой сильной связи между регионами
corr_abs = region_corr.abs()
np.fill_diagonal(corr_abs.values, 0)
print("\nСамая сильная связь между регионами:")
print(corr_abs.stack().idxmax(), "=", corr_abs.stack().max())
```

Сильнее всего связаны между собой продажи в Северной Америке и Европе (0.769), а также в Европе и остальных регионах (0.726). Слабее всего с другими регионами связаны продажи в Японии — особенно с остальными регионами (0.291). Это говорит о том, что японский рынок видеоигр имеет свою специфику и слабо предсказывает успех игры в других частях мира.
